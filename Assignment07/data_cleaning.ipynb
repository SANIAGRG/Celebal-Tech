```python
import pandas as pd

df = pd.read_csv("dataset/superstore_utf8.csv")
print("Loaded dataset with shape:", df.shape)
```

## 3. Handle missing values

```python
missing_counts = df.isnull().sum()
print("Missing values per column:")
print(missing_counts[missing_counts > 0])
```

```python
# Postal Code is numeric and has a handful of missing values -> fill with 0 (unknown)
# Any row missing a critical identifier (Order ID / Customer ID / Product ID) is dropped
df["Postal Code"] = df["Postal Code"].fillna(0)

critical_cols = ["Order ID", "Customer ID", "Product ID"]
before = len(df)
df = df.dropna(subset=critical_cols)
print(f"Dropped {before - len(df)} rows missing critical identifiers")
print("\nRemaining missing values:\n", df.isnull().sum().sum())
```

## 4. Basic operations: filter rows, select columns

```python
# Filter: orders from the 'West' region with Sales > 500
west_high_sales = df[(df["Region"] == "West") & (df["Sales"] > 500)]
print("West region orders with Sales > 500:", west_high_sales.shape[0])
west_high_sales.head()
```

```python
# Select a subset of columns
selected_cols = df[["Order ID", "Customer Name", "Category", "Sales", "Quantity", "Profit"]]
selected_cols.head()
```

## 5. Remove duplicates

```python
dup_count = df.duplicated().sum()
print("Duplicate rows found:", dup_count)
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

## 6. Create a derived column: `total_amount = price * quantity`

The dataset has no standalone unit-price column. `Sales` is the line-item total, so unit `price` is derived as `Sales / Quantity`, then `total_amount` is recomputed as `price * Quantity` to demonstrate the derived-column technique.

```python
df["price"] = df["Sales"] / df["Quantity"]
df["total_amount"] = df["price"] * df["Quantity"]
df[["Sales", "Quantity", "price", "total_amount"]].head()
```

## 7. Save the cleaned dataset

```python
df.to_csv("cleaned_superstore.csv", index=False)
print("Saved cleaned dataset to cleaned_superstore.csv with shape:", df.shape)
```
